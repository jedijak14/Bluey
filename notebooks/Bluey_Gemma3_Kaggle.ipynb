```python
import os
os.environ["KERAS_BACKEND"] = "torch"

import kagglehub

# Download latest version
path = kagglehub.model_download("keras/gemma3/keras/gemma3_1b")
print("Path to model files:", path)
```

```python
import tensorflow as tf
tf.config.set_visible_devices([], "GPU")
import keras
import keras_hub

keras.config.set_dtype_policy("float32")
gemma = keras_hub.models.Gemma3CausalLM.from_preset(path)
gemma.compile(sampler="greedy", jit_compile=False)
print("Requested preset loaded. No fine-tuning has been performed.")
```

```python
prompt = "Question: What is photosynthesis?\nAnswer:"
result = gemma.generate(prompt, max_length=160)
print(result)
```
